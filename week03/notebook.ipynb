## 2. Python Functions and Basic Operations

Let's start by understanding how to write Python functions with parameters and return values.

````python
# Example: A simple function with parameters
def greet(name):
    """Returns a greeting message."""
    return f"Hello, {name}!"

# Test the function
result = greet("Alice")
print(result)
````

````python
# Exercise: Write a function that adds two numbers
def add_numbers(a, b):
    """Returns the sum of two numbers."""
    return a + b

# Test it
print(f"5 + 3 = {add_numbers(5, 3)}")
````

## 3. String Manipulation and Mad Libs

Now let's create Mad Libs-style functions that combine user words into a story.

````python
# Example: String concatenation
adjective = "silly"
noun = "cat"
verb = "jumped"

# Simple concatenation
simple_story = f"The {adjective} {noun} {verb}."
print(simple_story)
````

````python
# Now let's look at the generate_mad_lib function we created
def generate_mad_lib(adjective, noun, verb):
    """Generates a short story using the provided words."""
    story = (
        f"On a {adjective} afternoon, a {noun} {verb} through the town, "
        f"making everyone smile as it passed by. People talked about the {noun} for days."
    )
    return story

# Test with different inputs
story1 = generate_mad_lib("silly", "cat", "jumped")
print("Story 1:")
print(story1)
print()

story2 = generate_mad_lib("brave", "knight", "battled")
print("Story 2:")
print(story2)
````

## 4. User Input and Conditional Logic

Let's explore how to use loops and conditionals to create interactive programs.

````python
# Example: Simple conditional
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"Score: {score}, Grade: {grade}")
````

````python
# Example: Simple while loop
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
````

````python
# The guessing_game function implementation
def guessing_game():
    """Plays a number guessing game with the user."""
    secret_number = random.randint(1, 100)
    attempts = 0

    print("Welcome to the Number Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")

    while True:
        attempts += 1
        guess_str = input("Enter your guess: ")
        try:
            guess = int(guess_str)
        except ValueError:
            print("Please enter a valid integer.")
            continue

        if guess < secret_number:
            print("Too low! Try again.")
        elif guess > secret_number:
            print("Too high! Try again.")
        else:
            print(f"Congratulations! You guessed it in {attempts} attempts!")
            break

# Note: The game is interactive - run it only when you want to play!
# To test it in this notebook, we'll use mocking (see next section)
````

## 5. Writing Unit Tests with pytest

Now let's write tests to verify our functions work correctly.

````python
# Test 1: Test the generate_mad_lib function
def test_generate_mad_lib():
    """Test that generate_mad_lib creates stories with all input words."""
    
    # Test case 1
    story = generate_mad_lib("silly", "cat", "jumped")
    
    # Assertions
    assert isinstance(story, str), "Function should return a string"
    assert "silly" in story, "Adjective not found in story"
    assert "cat" in story, "Noun not found in story"
    assert "jumped" in story, "Verb not found in story"
    
    print("✓ Test case 1 passed")
    
    # Test case 2
    story2 = generate_mad_lib("brave", "knight", "battled")
    assert "brave" in story2
    assert "knight" in story2
    assert "battled" in story2
    
    print("✓ Test case 2 passed")
    print("✓ All generate_mad_lib tests passed!")

# Run the test
test_generate_mad_lib()
````

````python
# Test 2: Test the guessing_game function with mocking
def test_guessing_game_with_mock():
    """Test guessing_game using mocks to control input/output."""
    
    # Mock the random.randint to return a predictable number
    with patch('random.randint', return_value=50):
        # Mock input to simulate user guesses
        with patch('builtins.input', side_effect=['75', '25', '50']):
            # Mock print to capture output
            with patch('builtins.print') as mock_print:
                guessing_game()
                
                # Verify print was called
                assert mock_print.called, "Game should produce output"
                print(f"✓ Game produced {mock_print.call_count} print statements")
                
    print("✓ Guessing game test passed!")

# Run the test
test_guessing_game_with_mock()
````

## 6. Running and Interpreting Test Results

Let's verify our functions are working correctly with a comprehensive test suite.

````python
# Summary of all tests
print("=" * 50)
print("SUMMARY: Running All Tests")
print("=" * 50)
print()

try:
    test_generate_mad_lib()
    print()
    test_guessing_game_with_mock()
    print()
    print("=" * 50)
    print("✓ ALL TESTS PASSED!")
    print("=" * 50)
except AssertionError as e:
    print(f"✗ Test failed: {e}")
````
